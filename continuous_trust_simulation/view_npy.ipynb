# Inspect Generated Anomaly Scores & Features
A simple utility notebook to load and plot the intermediate binary files (`.npy`) generated by the pipeline. This helps verify that the feature extraction and anomaly scoring steps ran correctly.

### 1. Anomaly Score Distribution (`anomaly_scores.npy`)
Loads the anomaly scores from `step3_anomaly_scoring.py` and plots a log-scaled histogram. Benign events should cluster very close to 0, while malicious events show up in the higher score ranges.

```python
import numpy as np
import matplotlib.pyplot as plt

arr = np.load('anomaly_scores.npy')
print(f"Shape: {arr.shape} | Dtype: {arr.dtype}")
print(f"Min: {arr.min():.4f} | Max: {arr.max():.4f} | Mean: {arr.mean():.4f}")

plt.hist(arr, bins=50, color='steelblue')
plt.title('Anomaly Score Distribution')
plt.xlabel('Score')
plt.ylabel('Count')
plt.yscale('log')
plt.show()
```

### 2. High-Dimensional Feature Space & Sparsity Inspection (`features.npy`)
Loads the 538-dimensional sparse feature matrix from `step2_extract_features.py`. Flattening and plotting the matrix on a log scale shows the high sparsity of the dataset (around 92.7% of the values are zeros, which is typical for AWS CloudTrail log features).

```python
import numpy as np
import matplotlib.pyplot as plt

arr = np.load('features.npy', allow_pickle=True)

# Info
print(f"Shape: {arr.shape}")
print(f"Dtype: {arr.dtype}")
print(f"Min: {arr.min():.4f} | Max: {arr.max():.4f} | Mean: {arr.mean():.4f}")
print(f"First 10 values: {arr[:10]}")

# Plot
plt.hist(arr.flatten(), bins=50, color='steelblue')
plt.title('Distribution')
plt.xlabel('Value')
plt.ylabel('Count')
plt.yscale('log')
plt.show()
```
