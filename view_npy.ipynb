# AI-Driven Behavioral Analytics & Feature Matrix Inspection
This notebook serves as an interactive utility to load, inspect, and visualize the intermediate binary arrays generated during our continuous trust re-evaluation pipeline (associated with **Paper 3**).

### 1. Anomaly Score Distribution (`anomaly_scores.npy`)
This cell loads the calibrated anomaly scores $s_u(t) \in [0, 1]$ generated by the unsupervised **Isolation Forest** model in `step3_anomaly_scoring.py`. It computes basic statistical metrics and plots a **log-scaled histogram** to visualize the distribution, confirming that the overwhelming majority of benign control-plane events concentrate near zero while malicious activities are pulled into elevated ranges.

```python
import numpy as np
import matplotlib.pyplot as plt

arr = np.load('anomaly_scores.npy')
print(f"Shape: {arr.shape} | Dtype: {arr.dtype}")
print(f"Min: {arr.min():.4f} | Max: {arr.max():.4f} | Mean: {arr.mean():.4f}")

plt.hist(arr, bins=50, color='steelblue')
plt.title('Anomaly Score Distribution')
plt.xlabel('Score')
plt.ylabel('Count')
plt.yscale('log')
plt.show()
```

### 2. High-Dimensional Feature Space & Sparsity Inspection (`features.npy`)
This cell loads the raw **538-dimensional sparse feature vectors** generated in `step2_extract_features.py` (representing one-hot categorical commands, log-scaled bytes, travel geodesics, and circadian hour offsets). It flattens the array and plots its value distribution to demonstrate the **extreme sparsity (7.3% density / 92.7% zeros)** characteristic of AWS CloudTrail control-plane audits.

```python
import numpy as np
import matplotlib.pyplot as plt

arr = np.load('features.npy', allow_pickle=True)

# Info
print(f"Shape: {arr.shape}")
print(f"Dtype: {arr.dtype}")
print(f"Min: {arr.min():.4f} | Max: {arr.max():.4f} | Mean: {arr.mean():.4f}")
print(f"First 10 values: {arr[:10]}")

# Plot
plt.hist(arr.flatten(), bins=50, color='steelblue')
plt.title('Distribution')
plt.xlabel('Value')
plt.ylabel('Count')
plt.yscale('log')
plt.show()
```
