# Inspect Generated Anomaly Scores
This utility notebook loads and inspects the binary `.npy` arrays generated by your Isolation Forest scorer (`step3_anomaly_scoring.py`). Use this to quickly verify and analyze the anomaly scores distribution.

```python
import numpy as np

# 1. Load the generated anomaly scores
scores = np.load("anomaly_scores.npy")

# 2. Print basic array metrics
print(f"Total scored events : {len(scores):,}")
print(f"Mean anomaly score  : {scores.mean():.4f}")
print(f"Max anomaly score   : {scores.max():.4f}")
print(f"Min anomaly score   : {scores.min():.4f}")
print(f"Sparsity (<0.10)     : {(scores < 0.10).mean():.1%} of events")

# 3. Display sample scores
print("\nSample Scores (First 10 events):")
print(scores[:10])
```
